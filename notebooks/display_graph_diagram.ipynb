# LangGraph Diagram Viewer

This notebook loads `docs/diagrams/graph_diagram.mmd` and renders it as Mermaid.

````python
from pathlib import Path
from IPython.display import Markdown, Image, display
import os
import base64
import requests

# Build robust candidate paths so notebook works from different working directories.
cwd = Path.cwd()
candidates = [
    cwd / "docs" / "diagrams" / "graph_diagram.mmd",
    cwd.parent / "docs" / "diagrams" / "graph_diagram.mmd",
    cwd.parent.parent / "docs" / "diagrams" / "graph_diagram.mmd",
    Path("notebooks") / ".." / "docs" / "diagrams" / "graph_diagram.mmd",
    Path(__file__).resolve().parent.parent / "docs" / "diagrams" / "graph_diagram.mmd" if "__file__" in globals() else None,
    Path(os.getcwd()).resolve().parent / "docs" / "diagrams" / "graph_diagram.mmd",
]

mmd_path = None
for candidate in candidates:
    if candidate is not None:
        resolved = candidate.resolve()
        if resolved.exists():
            mmd_path = resolved
            break

if mmd_path is None:
    raise FileNotFoundError(
        "Mermaid file not found. Checked:\n" + "\n".join(str(c.resolve()) for c in candidates if c is not None)
    )

mermaid_text = mmd_path.read_text(encoding="utf-8")
print(f"Using Mermaid file: {mmd_path}")

# 1) Try markdown-mermaid rendering (works in some notebook frontends)
display(Markdown(f"```mermaid\n{mermaid_text}\n```"))

# 2) Always try generating a real PNG image via Mermaid.ink
encoded = base64.urlsafe_b64encode(mermaid_text.encode("utf-8")).decode("ascii")
img_url = f"https://mermaid.ink/img/{encoded}"
print(f"Mermaid image URL: {img_url}")

try:
    resp = requests.get(img_url, timeout=20)
    resp.raise_for_status()
    png_bytes = resp.content

    out_dir = Path("reports")
    out_dir.mkdir(parents=True, exist_ok=True)
    out_file = out_dir / "graph_diagram.png"
    out_file.write_bytes(png_bytes)

    print(f"Saved PNG to: {out_file.resolve()}")
    display(Image(data=png_bytes))
except Exception as e:
    print("Could not fetch Mermaid PNG from mermaid.ink.")
    print(f"Reason: {e}")
    print("Open the URL above in a browser to render and download the image.")
````

````python
# Optional fallback: print Mermaid source if renderer is not available
if "mermaid_text" in globals():
    print(mermaid_text)
else:
    print("Run Cell 2 first to load the Mermaid file.")
````
